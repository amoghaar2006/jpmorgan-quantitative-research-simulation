```python
import pandas as pd
import numpy as np

# --- Load data ---
df = pd.read_csv("Task 3 and 4_Loan_Data.csv")

# --- Number of buckets ---
n_buckets = 5

# --- Create FICO buckets using quantiles ---
df['fico_bucket'] = pd.qcut(df['fico_score'], q=n_buckets, labels=False)

# --- Convert to rating (lower = better credit) ---
df['rating'] = n_buckets - df['fico_bucket']

# --- View results ---
print(df[['fico_score', 'rating']].head())


# --- Function to map new FICO score ---
# Save bucket boundaries
bins = pd.qcut(df['fico_score'], q=n_buckets, retbins=True)[1]

def map_fico_to_rating(fico_score):
    bucket = np.digitize(fico_score, bins, right=True) - 1
    rating = n_buckets - bucket
    return rating


# --- Test ---
print("FICO 750 → Rating:", map_fico_to_rating(750))
print("FICO 600 → Rating:", map_fico_to_rating(600))
```
